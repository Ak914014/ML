```python
def extract_episodes(txt):
    data = ""
    check = False
    for i in txt:
        if i == ")" :
            check = False
            return data
        if check == True :
            data = data + i 
        if i == "(" :
            check = True
            
```

```python
df["Episode"] = df["Title"].apply(extract_episodes)
```

```python
df["Episode"]= df["Episode"].str.replace(" eps","")
```

```python
df["Episode"] = df['Episode'].astype(int)
```

```python
def timeStamp (txt):
     data = ""
     check = False
     for i in range(len(txt)):
        if txt[i] == ")" :
            for j in range(i+1, i+20):
                data += txt[j]

            return data
```

```python
df['Title'].apply(timeStamp)
```

```python
df["TimeStamp"] = df['Title'].apply(timeStamp)
```

```python
df.info()
```

```python
df.head()
```

```python
def total_months(date_range):
    start_date, end_date = date_range.split(" - ")

    start = pd.Period(start_date.strip(), freq="M")
    end = pd.Period(end_date.strip(), freq="M")

    return end.ordinal - start.ordinal + 1
```

```python
df["TimeStamp"] = df['TimeStamp'].apply(total_months)
```

```python
df["TimeStamp"] = df['TimeStamp'].astype(int)
```
