```python
import numpy as np 
from scipy import stats 
```

```python
Group_A= [54.97, 48.61, 56.48, 65.23, 47.66, 47.66, 65.79, 42.33, 45.31, 55.43, 45.36, 45.35, 52.42, 30.87, 32.75]
Group_B = [48.24, 38.35, 51.23, 44.11, 72.33, 37.38, 51.13, 36.3 , 56.09, 58.74, 50.31, 46.12, 59.39, 44.07, 41.76]
```

```python
t_test, p_value = stats.ttest_ind(Group_A, Group_B , equal_var= False)
```

```python
t_test
```

```python
alpha = 0.05
```

```python
if p_value == alpha:
    print("i will reject the null hypo")
else :
    print("i will accept the null hypo")
```
