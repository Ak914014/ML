```python
import numpy as np 
arr = np.array([2,3,4,5,6,7,8,9,12,13,16,17,23,25,27,34,302])
arr
```

```python
q1 = np.percentile(arr, 25)
q3 = np.percentile(arr, 75)

q1+q3
```

```python
iqr = q3 - q1 
iqr 
```

```python
UF = q3 + 1.5*(iqr)
LF = q1 - 1.5*(iqr)
UF
```

```python
LF
```

```python
l = []
for i in arr:
    if i >= LF and i <= UF:
        l.append(i)
    else:
        break
        
        
```

```python
print(isinstance(l, list))
print(isinstance(True,int))
```

```python
arr2 = np.array(l)
arr2
```
